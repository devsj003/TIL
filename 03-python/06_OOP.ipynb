```python
class Circle:
    #생성자함수 -> 처음 인스턴스가 만들어질때 자동 호출되는 함수.
    def __init__(self, r):
        self.r = r
        print(f'반지름이 {self.r}인 원이 생성되었습니다.')

    def get_area(self):
        return 3.14 * self.r**2
    
    def __add__(self, other):
        return Circle(self.r + other.r)

c1 = Circle(3)
c2 = Circle(5)

print(c1.get_area())
print(c2.get_area())

c3 = c1 + c2
c3.get_area()
```
