```python
# Agent 내부의 Tool 활용해서 공통적으로 사용할 양식state을 생성

from langchain.agents import AgentState
# typing: 데이터의 형태를 설명하기 위함
from typing_extensions import NotRequired
from typing import Literal

# SupportStep 은 말 그대로 [a,b,c] 중 하나임
SupportStep = Literal['warrenty_collector', 'issue_checker', 'solution_maker']

class SupportSate(AgentState):
    """고객 지원 업무 흐름을 위한 State"""
    # Agent 내부에서 현재 어떤 단계를 진행중인지 확인용
    
    # 현재 단계
    current_step: NotRequired[SupportStep]
    # 보증 상태
    warranty_status: NotRequired[Literal['in_warranty', 'out_of_warranty']]
    # 이슈 타입
    issue_type: NotRequired[Literal['hw', 'sw']]
```
